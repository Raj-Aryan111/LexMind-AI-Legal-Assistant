```python
import pdfplumber
import re

from pydantic import BaseModel
from typing import Optional, Literal


class StatuteChunk(BaseModel):

    section_number: str
    act: Literal["BNS", "IPC"]
    chapter: str
    text: str

    bailable: Optional[bool] = None
    cognizable: Optional[bool] = None

    punishment_min_years: Optional[int] = None
    punishment_max_years: Optional[int] = None


def parse_statute_pdf(pdf_path: str, act: str):

    chunks = []

    current_section = None
    current_text = []

    current_chapter = "Unknown"

    section_pattern = re.compile(r"^(\d+)\.\s+(.+)")
    chapter_pattern = re.compile(
        r"^CHAPTER\s+[IVXLC]+\s*[-—]?\s*(.+)",
        re.IGNORECASE
    )

    with pdfplumber.open(pdf_path) as pdf:

        for page in pdf.pages:

            text = page.extract_text() or ""

            for line in text.splitlines():

                line = line.strip()

                ch = chapter_pattern.match(line)

                if ch:
                    current_chapter = ch.group(1).strip()
                    continue

                m = section_pattern.match(line)

                if m:

                    if current_section and current_text:

                        chunks.append(
                            StatuteChunk(
                                section_number=current_section,
                                act=act,
                                chapter=current_chapter,
                                text=" ".join(current_text)
                            )
                        )

                    current_section = m.group(1)

                    current_text = [line]

                elif current_section:
                    current_text.append(line)

    return chunks


bns_chunks = parse_statute_pdf(
    f"{BASE}/data/raw/statutes/BNS_2023.pdf",
    "BNS"
)

ipc_chunks = parse_statute_pdf(
    f"{BASE}/data/raw/statutes/IPC_1860.pdf",
    "IPC"
)

print(f"BNS Sections: {len(bns_chunks)}")
print(f"IPC Sections: {len(ipc_chunks)}")
```

```python
import pandas as pd

meta_df = pd.read_csv(
    f"{BASE}/data/raw/schedule1_metadata.csv"
)

meta = {
    (str(r.section_number), r.act): r
    for _, r in meta_df.iterrows()
}

for chunk in bns_chunks + ipc_chunks:

    key = (chunk.section_number, chunk.act)

    if key in meta:

        row = meta[key]

        chunk.bailable = bool(row.bailable)

        chunk.cognizable = bool(row.cognizable)

        chunk.punishment_min_years = (
            int(row.punishment_min)
            if pd.notna(row.punishment_min)
            else None
        )

        chunk.punishment_max_years = (
            int(row.punishment_max)
            if pd.notna(row.punishment_max)
            else None
        )

print("Metadata merged successfully")
```

```python
import fitz
import os
import json

def pdf_to_text(pdf_path):

    doc = fitz.open(pdf_path)

    text = "\n".join(
        page.get_text()
        for page in doc
    )

    return text


all_judgments = []

for root, _, files in os.walk("/content/sc_extracted/"):

    for fname in files:

        if not fname.endswith(".pdf"):
            continue

        path = os.path.join(root, fname)

        text = pdf_to_text(path)

        if len(text) < 500:
            continue

        all_judgments.append({
            "tid": fname.replace(".pdf", ""),
            "title": fname.replace(".pdf", ""),
            "doc": text,
            "docsource": "Supreme Court of India"
        })

print(f"Loaded {len(all_judgments)} judgments")
```

```python
from pydantic import BaseModel
from typing import List
import re


class JudgmentChunk(BaseModel):

    chunk_id: str
    parent_judgment_id: str

    court: str
    year: int

    petitioner: str
    respondent: str

    sections_cited: List[str]

    outcome: str

    block_type: str

    text: str


def split_blocks(full_text):

    text_lower = full_text.lower()

    blocks = {
        "headnote": "",
        "facts": "",
        "ratio": "",
        "obiter": "",
        "held": ""
    }

    # HEADNOTE
    blocks["headnote"] = full_text[:1500]

    # FACTS
    facts_match = re.search(
        r"(facts?.{0,50})(.*?)(held|judgment|analysis|reasoning)",
        text_lower,
        re.DOTALL
    )

    if facts_match:
        start = facts_match.start(2)
        end = facts_match.end(2)
        blocks["facts"] = full_text[start:end]

    # RATIO
    ratio_match = re.search(
        r"(reasoning|analysis|consideration)(.*?)(held|order|conclusion)",
        text_lower,
        re.DOTALL
    )

    if ratio_match:
        start = ratio_match.start(2)
        end = ratio_match.end(2)
        blocks["ratio"] = full_text[start:end]

    # HELD
    held_match = re.search(
        r"(held|order|conclusion)(.*)",
        text_lower,
        re.DOTALL
    )

    if held_match:
        start = held_match.start(2)
        blocks["held"] = full_text[start:]

    # OBITER
    blocks["obiter"] = full_text[-2000:]

    return blocks


def chunk_judgment(doc: dict) -> List[JudgmentChunk]:

    full_text = doc.get("doc", "")

    judgment_id = str(doc.get("tid", ""))

    court = (
        "Supreme Court"
        if "Supreme" in doc.get("docsource", "")
        else "High Court"
    )

    year_match = re.search(
        r"\b(19|20)\d{2}\b",
        doc.get("title", "")
    )

    year = (
        int(year_match.group())
        if year_match
        else 2000
    )

    sections = re.findall(
        r"[Ss]ection\s+(\d+[A-Z]?)\s+(?:IPC|BNS|CrPC|BNSS)",
        full_text
    )

    title = doc.get("title", "")

    if " v. " in title:
        parts = title.split(" v. ")
    elif " vs " in title:
        parts = title.split(" vs ")
    else:
        parts = [title, title]

    petitioner = parts[0][:100]
    respondent = parts[-1][:100]

    blocks = split_blocks(full_text)

    chunks = []

    for block_type, text in blocks.items():

        text = text.strip()

        if len(text) < 100:
            continue

        chunks.append(
            JudgmentChunk(
                chunk_id=f"{judgment_id}_{block_type}",
                parent_judgment_id=judgment_id,

                court=court,
                year=year,

                petitioner=petitioner,
                respondent=respondent,

                sections_cited=list(set(sections)),

                outcome="unknown",

                block_type=block_type,

                text=text
            )
        )

    return chunks


all_judgment_chunks = []

for doc in all_judgments:

    try:

        chunks = chunk_judgment(doc)

        all_judgment_chunks.extend(chunks)

    except Exception as e:

        print("Chunking failed:", e)

print(f"Created {len(all_judgment_chunks)} judgment chunks")
```

```python
from langchain_core.runnables import (

    RunnableLambda,

    RunnableParallel
)


# statute retrieval branch
statute_run = RunnableLambda(

    lambda x: statute_retriever.invoke(
        x["query"]
    )
)


# judgment retrieval branch
judgment_run = RunnableLambda(

    lambda x: judgment_retriever.invoke(
        x["query"]
    )
)


# mixed retrieval branch
parallel_run = RunnableParallel(

    statutes=statute_run,

    judgments=judgment_run
)


# simple manual router
def manual_router(x):

    q = x["query"].lower()


    # precedent / judgment queries
    if any(word in q for word in [

        "judgment",
        "case law",
        "precedent",
        "ruling",
        "supreme court",
        "high court"

    ]):

        return judgment_retriever.invoke(
            q
        )


    # mixed queries
    elif any(word in q for word in [

        "and",
        "along with",
        "also"

    ]) and any(word in q for word in [

        "punishment",
        "section",
        "ipc",
        "bns"

    ]):

        return {

            "statutes": statute_retriever.invoke(q),

            "judgments": judgment_retriever.invoke(q)
        }


    # default -> statutes
    else:

        return statute_retriever.invoke(
            q
        )


# final routing pipeline
full_routing = RunnableLambda(
    manual_router
)


print("Manual routing ready")
```

```python
from sentence_transformers import CrossEncoder

from langchain_core.runnables import (
    RunnableLambda
)


# CrossEncoder does query-document comparison directly
# much more accurate than embedding similarity
cross_encoder = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)


# rerank retrieved docs using CrossEncoder scores
def rerank(query: str, docs: list, top_k=3):

    # safety check
    if not docs:

        return docs


    # create (query, document) pairs
    pairs = [

        (query, d.page_content)

        for d in docs
    ]


    # CrossEncoder gives relevance score to each pair
    scores = cross_encoder.predict(
        pairs
    )


    # sort docs by score descending
    ranked = sorted(

        zip(scores, docs),

        key=lambda x: x[0],

        reverse=True
    )


    # keep only top_k best docs
    return [

        d

        for _, d in ranked[:top_k]
    ]


# wrap reranker as LCEL runnable
reranker_runnable = RunnableLambda(

    lambda x: rerank(

        x["query"],

        x["docs"]
    )
)


print("CrossEncoder reranker ready")
```

```python
from langchain_core.runnables import (
    RunnableLambda
)


# convert retrieved docs into readable context block
def context_formatter(docs) -> str:

    # mixed retrieval returns dict
    if isinstance(docs, dict):

        docs = (
            docs.get("statutes", [])
            +
            docs.get("judgments", [])
        )


    lines = []


    for d in docs:

        meta = d.metadata


        # statute formatting
        if "section_number" in meta:

            lines.append(

                f"[STATUTE: {meta['act']} §{meta['section_number']}]\n"

                f"{d.page_content}\n"
            )


        # judgment formatting
        else:

            lines.append(

                f"[JUDGMENT: "

                f"{meta.get('court','?')} "

                f"{meta.get('year','?')}]"

                f"\n{d.page_content}\n"
            )


    return "\n".join(lines)


# full retrieval pipeline
retrieval_chain = (

    RunnableLambda(

        lambda x: {

            "query": x["query"]
        }
    )

    | full_routing

    | RunnableLambda(context_formatter)
)


print("Full retrieval chain ready")


# end-to-end test
ctx = retrieval_chain.invoke({

    "query":
    "What BNS sections cover robbery with grievous hurt?"
})


print(ctx[:1000])
```

```python
# -------------------------------
# retrieval evaluation framework
# -------------------------------

from collections import defaultdict


# precision@k metric
def precision_at_k(retrieved, expected, k=3):

    retrieved_topk = retrieved[:k]

    hits = sum(
        1 for r in retrieved_topk
        if r in expected
    )

    return hits / k


# -------------------------------
# statute evaluation dataset
# -------------------------------

# query -> expected section numbers
statute_test_cases = [

    {
        "query":
        "punishment for murder",

        "expected":
        ["302", "103"]
    },

    {
        "query":
        "attempt to commit robbery with deadly weapon",

        "expected":
        ["398", "312"]
    },

    {
        "query":
        "kidnapping child under ten years",

        "expected":
        ["369", "97"]
    },

    {
        "query":
        "dowry death punishment",

        "expected":
        ["304B"]
    },

    {
        "query":
        "rape punishment",

        "expected":
        ["376"]
    },

    {
        "query":
        "criminal conspiracy",

        "expected":
        ["120B"]
    },

    {
        "query":
        "culpable homicide not amounting to murder",

        "expected":
        ["304"]
    },

    {
        "query":
        "theft punishment",

        "expected":
        ["378", "379"]
    },

    {
        "query":
        "cheating and dishonestly inducing delivery of property",

        "expected":
        ["420"]
    },

    {
        "query":
        "attempt to murder",

        "expected":
        ["307"]
    }
]


# -------------------------------
# statute evaluation
# -------------------------------

statute_scores = []


for test in statute_test_cases:

    query = test["query"]

    expected = test["expected"]


    docs = statute_retriever.invoke(
        query
    )


    retrieved_sections = []


    for d in docs:

        text = d.page_content[:100]


        # extract section number
        sec = text.split(".")[0].strip()

        retrieved_sections.append(sec)


    p3 = precision_at_k(

        retrieved_sections,

        expected,

        k=3
    )


    statute_scores.append(p3)


    print("\n" + "=" * 80)

    print("QUERY:", query)

    print("EXPECTED:", expected)

    print("RETRIEVED:", retrieved_sections[:3])

    print(f"Precision@3 = {p3:.2f}")


# -------------------------------
# final statute score
# -------------------------------

avg_statute_p3 = (
    sum(statute_scores)
    / len(statute_scores)
)

print("\n" + "=" * 80)

print(
    f"\nAverage Statute Precision@3: "
    f"{avg_statute_p3:.3f}"
)
```

```python
# -----------------------------------
# reranked retrieval evaluation
# -----------------------------------

from collections import defaultdict


# helper:
# extract section number from retrieved statute
def extract_section(doc):

    try:

        sec = (
            doc.page_content
            .split(".")[0]
            .strip()
        )

        return sec

    except:

        return "UNKNOWN"


# precision@k metric
def precision_at_k(retrieved, expected, k=3):

    retrieved_topk = retrieved[:k]

    hits = sum(

        1 for r in retrieved_topk

        if r in expected
    )

    return hits / k


# -----------------------------------
# reranked evaluation
# -----------------------------------

reranked_scores = []


for test in statute_test_cases:

    query = test["query"]

    expected = test["expected"]


    # first-stage retrieval
    docs = statute_retriever.invoke(
        query
    )


    # second-stage reranking
    reranked_docs = rerank(

        query=query,

        docs=docs,

        top_k=3
    )


    # extract retrieved sections
    retrieved_sections = [

        extract_section(d)

        for d in reranked_docs
    ]


    # compute precision
    p3 = precision_at_k(

        retrieved_sections,

        expected,

        k=3
    )


    reranked_scores.append(p3)


    print("\n" + "=" * 80)

    print("QUERY:", query)

    print("EXPECTED:", expected)

    print("RERANKED:", retrieved_sections)

    print(f"Precision@3 = {p3:.2f}")


# -----------------------------------
# final reranked score
# -----------------------------------

avg_reranked_p3 = (

    sum(reranked_scores)

    / len(reranked_scores)
)


print("\n" + "=" * 80)

print(

    f"\nRERANKED Average Precision@3: "
    f"{avg_reranked_p3:.3f}"
)


# compare before vs after
print("\n" + "=" * 80)

print(
    f"\nBaseline Precision@3 : "
    f"{avg_statute_p3:.3f}"
)

print(
    f"Reranked Precision@3: "
    f"{avg_reranked_p3:.3f}"
)

print(
    f"Improvement           : "
    f"{avg_reranked_p3 - avg_statute_p3:.3f}"
)
```

```python
from rank_bm25 import BM25Okapi


# pull all statutes from Qdrant
all_statute_docs = statute_store.similarity_search(
    "criminal law",
    k=2000
)


# tokenize text
bm25_corpus = [
    d.page_content.lower().split()
    for d in all_statute_docs
]


# BM25 keyword retriever
bm25 = BM25Okapi(bm25_corpus)

print(f"BM25 corpus size: {len(bm25_corpus)}")
```

```python
# semantic + keyword retrieval

def hybrid_retrieve(query, top_k=8):

    # semantic retrieval
    semantic_docs = statute_retriever.invoke(query)


    # BM25 retrieval
    tokenized_query = query.lower().split()

    bm25_scores = bm25.get_scores(tokenized_query)


    # top BM25 indices
    top_indices = sorted(
        range(len(bm25_scores)),
        key=lambda i: bm25_scores[i],
        reverse=True
    )[:top_k]


    bm25_docs = [
        all_statute_docs[i]
        for i in top_indices
    ]
 # merge both retrievals
    merged = semantic_docs + bm25_docs


    # remove duplicates
    seen = set()
    unique_docs = []


    for d in merged:

        text_key = d.page_content[:150]

        if text_key not in seen:

            seen.add(text_key)

            unique_docs.append(d)


    return unique_docs
```

```python
# simplified mapping table
IPC_TO_BNS = {

    "302": "103",

    "307": "109",

    "120B": "61",

    "304": "105",

    "376": "64",

    "420": "318",

    "379": "303",

    "398": "312",

    "369": "97"
}


# reverse mapping
BNS_TO_IPC = {
    v: k
    for k, v in IPC_TO_BNS.items()
}
```

```python
# smarter legal equivalence matcher

def legal_match(retrieved, expected):

    # exact match
    if retrieved == expected:

        return True


    # IPC -> BNS
    if IPC_TO_BNS.get(expected) == retrieved:

        return True


    # BNS -> IPC
    if BNS_TO_IPC.get(expected) == retrieved:

        return True


    return False
```

```python
# legal-aware precision metric

def legal_precision_at_k(retrieved, expected, k=3):

    retrieved_topk = retrieved[:k]


    hits = 0


    for r in retrieved_topk:

        for e in expected:

            if legal_match(r, e):

                hits += 1

                break


    return hits / k
```

```python
hybrid_scores = []


for test in statute_test_cases:

    query = test["query"]

    expected = test["expected"]


    # hybrid retrieval
    docs = hybrid_retrieve(query)


    # CrossEncoder reranking
    reranked_docs = rerank(
        query=query,
        docs=docs,
        top_k=3
    )


    # extract sections
    retrieved_sections = []


    for d in reranked_docs:

        sec = (
            d.page_content
            .split(".")[0]
            .strip()
        )

        retrieved_sections.append(sec)


    # improved metric
    p3 = legal_precision_at_k(
        retrieved_sections,
        expected,
        k=3
    )


    hybrid_scores.append(p3)


    print("\n" + "=" * 80)

    print("QUERY:", query)

    print("EXPECTED:", expected)

    print("RETRIEVED:", retrieved_sections)

    print(f"Precision@3 = {p3:.2f}")


# final score
avg_hybrid_p3 = (
    sum(hybrid_scores)
    / len(hybrid_scores)
)


print("\n" + "=" * 80)

print(
    f"\nHybrid + Reranked Precision@3: "
    f"{avg_hybrid_p3:.3f}"
)
```

```python
# ---------------------------------------------------
# better formatter
# ---------------------------------------------------

def context_formatter(docs) -> str:


    # mixed retrieval handling
    if isinstance(docs, dict):

        docs = (

            docs.get("statutes", [])

            +

            docs.get("judgments", [])
        )


    lines = []


    for d in docs:

        meta = d.metadata

        text = d.page_content.strip()


        # statute formatting
        if (

            meta.get("_collection_name")

            == "statutes"
        ):

            section = (

                text.split(".")[0]

                .strip()
            )


            lines.append(

                f"[STATUTE §{section}]\n"

                f"{text}\n"
            )


        # judgment formatting
        else:

            lines.append(

                f"[JUDGMENT: "

                f"{meta.get('court','?')} "

                f"{meta.get('year','?')}]\n"

                f"{text}\n"
            )


    return "\n".join(lines)


print("Formatter updated")
```

```python
from pydantic import (

    BaseModel,

    Field
)

from typing import (

    List,

    Optional
)


# ===================================================
# STATUTE CITATION MODEL
# ===================================================

class StatuteCitation(BaseModel):

    section_number: str

    act: str

    section_title: str

    relevance_explanation: str

    # exact retrieved chunk
    retrieved_text: str



# ===================================================
# JUDGMENT CITATION MODEL
# ===================================================

class JudgmentCitation(BaseModel):

    case_name: str

    court: str

    year: int

    headnote: str

    how_it_applies: str

    # exact retrieved chunk
    retrieved_text: str



# ===================================================
# FINAL LEGAL RESPONSE MODEL
# ===================================================

class LegalResponse(BaseModel):

    # final synthesized answer
    summary: str


    # statute citations
    cited_statutes: List[
        StatuteCitation
    ]


    # precedent citations
    cited_judgments: List[
        JudgmentCitation
    ]


    # retrieval confidence
    confidence_score: float


    # optional warnings
    caveat: Optional[str] = None



# ===================================================
# ANTI-HALLUCINATION VALIDATION
# ===================================================

def validate_citations(

    response: LegalResponse

) -> LegalResponse:


    valid_statutes = []


    for cit in response.cited_statutes:


        # verify statute exists in Qdrant
        hits = client.scroll(

            collection_name="statutes",

            scroll_filter={

                "must": [

                    {

                        "key":
                        "section_number",

                        "match": {

                            "value":
                            cit.section_number
                        }
                    }
                ]
            },

            limit=1
        )[0]


        # keep valid citations
        if hits:

            valid_statutes.append(
                cit
            )


        # remove hallucinated citations
        else:

            if response.caveat is None:

                response.caveat = ""


            response.caveat += (

                f" [REMOVED hallucinated "
                f"§{cit.section_number}]"
            )


    # overwrite with verified statutes
    response.cited_statutes = (
        valid_statutes
    )


    return response


print("LegalResponse schema ready")
```

```python
# ===================================================
# STRICT EXACT SECTION VALIDATION
# ===================================================

import re

from qdrant_client.http.models import (

    Filter,

    FieldCondition,

    MatchValue
)


# ===================================================
# EXACT SECTION LOOKUP
# ===================================================

def exact_section_lookup(

    query: str
):


    # detect patterns like:
    # 302 IPC
    # Section 103 BNS
    # 304B IPC

    pattern = (

        r"(?:Section|section)?"

        r"\s*"

        r"(\d+[A-Z]?)"

        r"\s*"

        r"(IPC|BNS)"
    )


    match = re.search(

        pattern,

        query
    )


    # no explicit section found
    if not match:

        return None


    section_number = (
        match.group(1)
    )

    act = (
        match.group(2)
    )


    print(

        f"\nChecking exact section: "
        f"{section_number} {act}"
    )


    # exact Qdrant filter
    qdrant_filter = Filter(

        must=[

            FieldCondition(

                key="section_number",

                match=MatchValue(

                    value=section_number
                )
            ),

            FieldCondition(

                key="act",

                match=MatchValue(

                    value=act
                )
            )
        ]
    )


    hits = client.scroll(

        collection_name="statutes",

        scroll_filter=qdrant_filter,

        limit=1
    )[0]


    # exact section NOT found
    if not hits:

        return (

            "No relevant source found "
            "in database."
        )


    # valid section found
    payload = hits[0].payload


    return (

        f"[SECTION LOOKUP]\n\n"

        f"{payload.get('section_number')}."

        f" {payload.get('text')}"
    )


print("Exact section validator ready")


# ===================================================
# SAFE LOOKUP TOOL
# ===================================================

@tool
def lookup_section_details(

    query: str
) -> str:


    """
    Exact lookup for IPC/BNS sections.
    Prevents hallucinated section retrieval.
    """


    # -------------------------------------------
    # STEP 1 — STRICT EXACT LOOKUP
    # -------------------------------------------

    exact_result = exact_section_lookup(

        query
    )


    # exact lookup handled query
    if exact_result is not None:

        return exact_result


    # -------------------------------------------
    # STEP 2 — FALLBACK
    # -------------------------------------------

    return (

        "No relevant source found "
        "in database."
    )


print("Safe lookup_section_details tool ready")
```

```python
# ===================================================
# STRICT INVALID SECTION HANDLING
# ===================================================

@tool
def lookup_section_details(

    query: str
) -> str:


    """
    Exact lookup for IPC/BNS sections.
    Prevents hallucinated section retrieval.
    """


    # -------------------------------------------
    # EXACT LOOKUP
    # -------------------------------------------

    exact_result = exact_section_lookup(

        query
    )


    # exact section found
    if exact_result is not None:


        # invalid section
        if (

            "No relevant source found"
            in exact_result
        ):

            return (

                "[INVALID_SECTION]\n"

                "No relevant source found "
                "in database."
            )


        # valid section
        return exact_result


    # -------------------------------------------
    # NO SECTION DETECTED
    # -------------------------------------------

    return (

        "[INVALID_SECTION]\n"

        "No relevant source found "
        "in database."
    )


print("Strict lookup_section_details ready")
```
